```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image_path='images8.png'
img=cv2.imread(image_path,cv2.IMREAD_GRAYSCALE)

# BINARY THRESHOLD
_, binary_image=cv2.threshold(img,127,255,cv2.THRESH_BINARY_INV)

kernel=np.ones((8,8),np.uint8)
```

```python
#Dialation
dilated_image=cv2.dilate(binary_image,kernel,iterations=1)
plt.imshow(dilated_image,cmap='gray')
plt.show()
```

```python
#Erosion
eroded_image=cv2.erode(binary_image,kernel,iterations=1)
plt.imshow(eroded_image, cmap='gray')
```

```python
#opend
opened_image=cv2.morphologyEx(binary_image,cv2.MORPH_OPEN,kernel)
plt.imshow(opened_image,cmap='gray')
```

```python

#close
closed_image=cv2.morphologyEx(binary_image,cv2.MORPH_CLOSE,kernel)
plt.imshow(closed_image,cmap='gray')
```

```python
#gradient
gradient_image=cv2.morphologyEx(binary_image,cv2.MORPH_GRADIENT,kernel)
plt.imshow(gradient_image,cmap='gray')
```

```python
#Tophat
tophat_image=cv2.morphologyEx(binary_image,cv2.MORPH_TOPHAT,kernel)
plt.imshow(tophat_image,cmap='gray')
```

```python
#Blackhat
blackhat_image=cv2.morphologyEx(binary_image,cv2.MORPH_BLACKHAT,kernel)
plt.imshow(blackhat_image,cmap='gray')
```

```python
titles = ['Original', 'Dilation', 'Erosion', 'Opening', 'Closing', 'Gradient','Tophat','blackhat']
images = [img, dilated_image, eroded_image, opened_image, closed_image, gradient_image,tophat_image,blackhat_image]

plt.figure(figsize=(10, 7))

for i in range(8):
    plt.subplot(2, 4, i + 1)
    plt.imshow(images[i], cmap='gray')
    plt.title(titles[i])
    plt.axis('off')

plt.tight_layout()
plt.show()
```
